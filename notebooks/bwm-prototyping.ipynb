```python
import sys
from pathlib import Path
import pandas as pd

pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

root = str(Path.cwd().parent)

if root not in sys.path:
    sys.path.append(root)


from src.step3a_bwm_model import BWMCalculator
```

```python
# ==========================================
# 2. DECISION DEFINITION (User Input)
# ==========================================
criteria = ["cost", "engine", "airbag", "fuel_economy", "aesthetics"]

# The user chooses the Most Important and Least Important criteria
best_criterion = "cost"
worst_criterion = "aesthetics"

# The user compares the Best (Cost) against all others (Score from 1 to 9)
# 1 = Equal importance / 9 = Absolutely more important
bo_vector = {
    "engine": 3,         # Cost is 3x more important than Engine
    "airbag": 2,         # Cost is 2x more important than Airbag
    "fuel_economy": 4,   # Cost is 4x more important than Fuel Economy
    "aesthetics": 8      # Cost is 8x more important than Aesthetics
}

# The user compares all others against the Worst (Aesthetics) (Score from 1 to 9)
ow_vector = {
    "cost": 8,           # (This should reflect the score above to maintain consistency)
    "engine": 5,         # Engine is 5x more important than Aesthetics
    "airbag": 6,         # Airbag is 6x more important than Aesthetics
    "fuel_economy": 3    # Fuel Economy is 3x more important than Aesthetics
}

```

```python
# ==========================================
# 3. BWM ENGINE EXECUTION
# ==========================================
bwm_calculator = BWMCalculator(criteria)
weights, consistency_ratio = bwm_calculator.calculate_weights(best_criterion, worst_criterion, bo_vector, ow_vector)
```

```python
# ==========================================
# 4. RESULTS VISUALIZATION
# ==========================================
print(f"📊 Consistency Ratio (CR): {consistency_ratio}")
if consistency_ratio < 0.1:
    print("✅ Excellent! Highly consistent vectors.")
elif consistency_ratio < 0.2:
    print("⚠️ Acceptable. There is some inconsistency, but the result is valid.")
else:
    print("❌ Inconsistent! The user needs to review their 1-9 scores.")

# Transform the weights dictionary into a DataFrame for a beautiful Jupyter render
df_weights = pd.DataFrame(list(weights.items()), columns=['Criterion', 'BWM Weight'])
df_weights = df_weights.sort_values(by='BWM Weight', ascending=False).reset_index(drop=True)

# Format the visual as percentages with a blue color gradient
display(df_weights)
```
